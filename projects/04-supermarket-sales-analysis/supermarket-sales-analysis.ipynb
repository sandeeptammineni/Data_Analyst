``````````````````````````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
``````````````````````````

``````````````````````````python
df=pd.read_excel(r"C:\Users\harip\Downloads\archive (31)\supermarket_sales.xlsx")
df.head()
``````````````````````````

``````````````````````````python
df.isna().sum()
``````````````````````````

``````````````````````````python
df['Date']=pd.to_datetime(df['Date'])
df.head()
``````````````````````````

``````````````````````````python
df['year']=df['Date'].dt.year
df['month']=df['Date'].dt.month
df.head()
``````````````````````````

``````````````````````````python
df['Revenue']=df['Total']
df['Profit']=df['gross income']
df.drop(columns=['Total'],inplace=True)
df.head()
``````````````````````````

``````````````````````````python
df['Revenue']=round(df['Revenue']*100)
df['profit']=round(df['Profit']*100)
df.drop(columns=['Profit'],inplace=True)
``````````````````````````

``````````````````````````python
df.head()
``````````````````````````

``````````````````````````python
monthly_profit=df.groupby('month')['profit'].sum()
``````````````````````````

``````````````````````````python
monthly_profit.plot(kind='bar',color='skyblue')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.show()
``````````````````````````

``````````````````````````python
df.groupby('year')['profit'].sum()
``````````````````````````

``````````````````````````python
product_line_sales=df.groupby('Product line')['Quantity'].sum()
product_line_sales.plot(kind='bar',color='lightpink')
plt.title('Highest Selling Category')
plt.xlabel('Category')
plt.ylabel('Quantity')
plt.show()
``````````````````````````

``````````````````````````python
city_sales=df.groupby('City')['Revenue'].sum()
city_sales.plot(kind='bar',color='lightgreen')
plt.title('City wise Sales')
plt.xlabel('City')
plt.ylabel('Revenue generated (Cr)')
plt.show()
``````````````````````````

``````````````````````````python
avg_ratings = df.groupby('Product line')['Rating'].mean().sort_values(ascending=False)

# Plot
ax = avg_ratings.plot(kind='bar', color='skyblue', edgecolor='black', figsize=(8, 5))

# Add values on top of bars
for i, v in enumerate(avg_ratings):
    ax.text(i, v + 0.02, f'{v:.2f}', ha='center', va='bottom', fontweight='bold')

plt.title('Average Rating by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.7)
plt.tight_layout()

plt.show()
``````````````````````````

``````````````````````````python
payment_methods=df.groupby('Gender')['Revenue'].sum()
payment_methods.plot(kind='bar',color=['pink','black'])
plt.title('Most used Payment method')
plt.xlabel('Method')
plt.ylabel('total count')
plt.show()
``````````````````````````
